# 1. Project Introduction and Business Problem

## Project Overview
This project analyzes Walmart's historical weekly sales data for 45 stores from February 2010 to October 2012. The objective is to identify sales trends, understand factors associated with weekly sales, compare store performance, and forecast sales for the next 12 weeks.

## Business Problem
Retail businesses need reliable sales insights to plan inventory, manage resources, and prepare for seasonal demand. This project uses historical Walmart sales data to examine how sales vary across stores and time periods and how factors such as holidays, temperature, fuel prices, CPI, and unemployment relate to sales.

## Project Objectives
- Clean and validate the dataset.
- Perform exploratory data analysis and identify potential outliers.
- Analyze holiday and seasonal sales patterns.
- Investigate relationships between sales and economic or environmental factors.
- Identify the best- and worst-performing stores.
- Compare forecasting models and predict weekly sales for all 45 stores for the next 12 weeks.

## Tools and Technologies
Python, Pandas, NumPy, Matplotlib, Seaborn, and Statsmodels in Google Colab.

## Expected Outcome
The project aims to deliver clear visualizations, store performance comparisons, forecast evaluations, and practical recommendations to support data-driven retail planning.

```python

# ============================================
# WALMART CAPSTONE PROJECT
# ============================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Load the Walmart dataset
file_path = "/content/Walmart DataSet.csv"
df = pd.read_csv(file_path)

# Display basic information
print("WALMART DATASET OVERVIEW")
print("-" * 40)

print("Dataset Shape:", df.shape)

print("\nFirst 5 Rows:")
display(df.head())

print("\nColumn Names:")
print(df.columns.tolist())

print("\nDataset Information:")
df.info()

print("\nMissing Values:")
display(df.isnull().sum().to_frame(name="Missing Count"))

print("\nDuplicate Rows:", df.duplicated().sum())
```

```python
# ============================================

# Work on a copy of the original dataset
walmart = df.copy()

# Convert Date to datetime (day-month-year format)
walmart["Date"] = pd.to_datetime(
    walmart["Date"],
    format="%d-%m-%Y",
    errors="coerce"
)

# Check date conversion
print("Invalid dates:", walmart["Date"].isna().sum())

# Check data types after conversion
print("\nUpdated Data Types:")
print(walmart.dtypes)

# Check duplicate records
print("\nDuplicate Rows:", walmart.duplicated().sum())

# Check unique stores
print("\nNumber of Stores:", walmart["Store"].nunique())
print("Store IDs:", sorted(walmart["Store"].unique()))

# Check weekly observations per store
print("\nObservations per Store:")
display(walmart.groupby("Store").size().describe())

# Check date range
print("\nDate Range:")
print("Start Date:", walmart["Date"].min().date())
print("End Date:", walmart["Date"].max().date())

# Check negative and zero weekly sales
print("\nZero Sales Records:", (walmart["Weekly_Sales"] == 0).sum())
print("Negative Sales Records:", (walmart["Weekly_Sales"] < 0).sum())

# Summary statistics
print("\nSummary Statistics:")
display(walmart.describe().round(2))
```

```python

sns.set_theme(style="whitegrid", palette="deep")

# Ensure records are sorted chronologically within each store
walmart = walmart.sort_values(["Store", "Date"]).copy()

# Create date-related features
walmart["Year"] = walmart["Date"].dt.year
walmart["Month"] = walmart["Date"].dt.month
walmart["Week"] = walmart["Date"].dt.isocalendar().week.astype(int)

# 1. Overall weekly sales trend
overall_sales = (
    walmart.groupby("Date", as_index=False)["Weekly_Sales"]
    .sum()
    .sort_values("Date")
)

plt.figure(figsize=(14, 5))
sns.lineplot(
    data=overall_sales,
    x="Date",
    y="Weekly_Sales"
)
plt.title("Walmart: Total Weekly Sales Across All Stores")
plt.xlabel("Date")
plt.ylabel("Total Weekly Sales")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()


# 2. Sales distribution across stores
plt.figure(figsize=(14, 6))
sns.boxplot(
    data=walmart,
    x="Store",
    y="Weekly_Sales"
)
plt.title("Weekly Sales Distribution by Store")
plt.xlabel("Store ID")
plt.ylabel("Weekly Sales")
plt.xticks(rotation=90)
plt.tight_layout()
plt.show()


# 3. Average weekly sales by month
monthly_sales = (
    walmart.groupby("Month", as_index=False)["Weekly_Sales"]
    .mean()
)

plt.figure(figsize=(11, 5))
sns.barplot(
    data=monthly_sales,
    x="Month",
    y="Weekly_Sales"
)
plt.title("Average Weekly Sales by Calendar Month")
plt.xlabel("Month")
plt.ylabel("Average Weekly Sales")
plt.tight_layout()
plt.show()


# 4. Holiday vs non-holiday sales
holiday_sales = (
    walmart.groupby("Holiday_Flag")["Weekly_Sales"]
    .agg(["mean", "median", "count"])
    .rename(index={0: "Non-Holiday", 1: "Holiday"})
)

print("Holiday vs Non-Holiday Sales Summary")
display(holiday_sales.round(2))

plt.figure(figsize=(8, 5))
sns.boxplot(
    data=walmart,
    x="Holiday_Flag",
    y="Weekly_Sales"
)
plt.title("Weekly Sales: Holiday vs Non-Holiday")
plt.xlabel("Holiday Flag (0 = Non-Holiday, 1 = Holiday)")
plt.ylabel("Weekly Sales")
plt.tight_layout()
plt.show()
```

### Seasonal Sales Trends and Business Interpretation

The exploratory analysis indicates that Walmart weekly sales tend to increase during major holiday periods, particularly toward the end of the year. Holiday weeks may experience higher customer demand because of seasonal shopping, holiday preparations, promotions, and increased purchasing activity.

The holiday versus non-holiday comparison in this dataset shows average weekly sales of approximately $1,122,888 during holiday weeks and $1,041,256 during non-holiday weeks, a difference of approximately 7.84%.

These results suggest that holiday periods are important for sales planning. However, the observed difference does not prove that holidays alone cause higher sales. Store characteristics, promotions, economic conditions, and the specific holiday period may also influence sales.


```python
from scipy import stats

# --------------------------------------------
# Outlier detection using IQR per store
# --------------------------------------------

store_outlier_results = []

for store_id, group in walmart.groupby("Store"):
    q1 = group["Weekly_Sales"].quantile(0.25)
    q3 = group["Weekly_Sales"].quantile(0.75)
    iqr = q3 - q1

    lower_bound = q1 - 1.5 * iqr
    upper_bound = q3 + 1.5 * iqr

    outliers = group[
        (group["Weekly_Sales"] < lower_bound) |
        (group["Weekly_Sales"] > upper_bound)
    ]

    store_outlier_results.append({
        "Store": store_id,
        "Lower_Bound": lower_bound,
        "Upper_Bound": upper_bound,
        "Outlier_Count": len(outliers),
        "Outlier_Percentage": len(outliers) / len(group) * 100
    })

outlier_summary = pd.DataFrame(store_outlier_results)

print("OUTLIER SUMMARY BY STORE")
display(
    outlier_summary.sort_values(
        "Outlier_Count", ascending=False
    ).head(10).round(2)
)

print(
    "\nTotal potential sales outliers:",
    outlier_summary["Outlier_Count"].sum()
)

# --------------------------------------------
# Holiday vs non-holiday statistical test
# --------------------------------------------

non_holiday = walmart.loc[
    walmart["Holiday_Flag"] == 0, "Weekly_Sales"
]

holiday = walmart.loc[
    walmart["Holiday_Flag"] == 1, "Weekly_Sales"
]

# Welch's t-test does not assume equal variances
t_stat, p_value = stats.ttest_ind(
    holiday,
    non_holiday,
    equal_var=False
)

print("\nHOLIDAY SALES SIGNIFICANCE TEST")
print(f"Mean non-holiday sales: {non_holiday.mean():,.2f}")
print(f"Mean holiday sales:     {holiday.mean():,.2f}")
print(f"Difference:             {holiday.mean() - non_holiday.mean():,.2f}")
print(f"T-statistic:            {t_stat:.4f}")
print(f"P-value:                {p_value:.6f}")

if p_value < 0.05:
    print("Result: Statistically significant difference at the 5% level.")
else:
    print("Result: No statistically significant difference at the 5% level.")

# --------------------------------------------
# Overall correlation matrix
# --------------------------------------------

variables = [
    "Weekly_Sales",
    "Temperature",
    "Fuel_Price",
    "CPI",
    "Unemployment"
]

correlation_matrix = walmart[variables].corr()

print("\nCORRELATION MATRIX")
display(correlation_matrix.round(3))

plt.figure(figsize=(9, 6))
sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap="coolwarm",
    center=0,
    fmt=".2f"
)
plt.title("Correlation Between Weekly Sales and Economic Factors")
plt.tight_layout()
plt.show()

# --------------------------------------------
# Store-wise unemployment relationship
# --------------------------------------------

unemployment_results = []

for store_id, group in walmart.groupby("Store"):
    corr = group["Weekly_Sales"].corr(group["Unemployment"])

    unemployment_results.append({
        "Store": store_id,
        "Unemployment_Correlation": corr
    })

unemployment_summary = pd.DataFrame(unemployment_results)

print("\nSTORES WITH MOST NEGATIVE UNEMPLOYMENT CORRELATION")
display(
    unemployment_summary.sort_values(
        "Unemployment_Correlation"
    ).head(10).round(3)
)

print(
    "\nMedian store-wise unemployment correlation:",
    round(unemployment_summary["Unemployment_Correlation"].median(), 3)
)
```

```python

# --------------------------------------------
# Store-wise Temperature and CPI Analysis
# --------------------------------------------

store_factor_results = []

for store_id, group in walmart.groupby("Store"):
    store_factor_results.append({
        "Store": store_id,
        "Temperature_Correlation":
            group["Weekly_Sales"].corr(group["Temperature"]),
        "CPI_Correlation":
            group["Weekly_Sales"].corr(group["CPI"])
    })

store_factor_summary = pd.DataFrame(store_factor_results)

print("STORES WITH MOST NEGATIVE TEMPERATURE CORRELATION")
display(
    store_factor_summary.sort_values(
        "Temperature_Correlation"
    ).head(10).round(3)
)

print("STORES WITH MOST POSITIVE TEMPERATURE CORRELATION")
display(
    store_factor_summary.sort_values(
        "Temperature_Correlation", ascending=False
    ).head(10).round(3)
)

print("STORES WITH MOST NEGATIVE CPI CORRELATION")
display(
    store_factor_summary.sort_values(
        "CPI_Correlation"
    ).head(10).round(3)
)

print("STORES WITH MOST POSITIVE CPI CORRELATION")
display(
    store_factor_summary.sort_values(
        "CPI_Correlation", ascending=False
    ).head(10).round(3)
)

print("\nMEDIAN STORE-WISE CORRELATIONS")
print(
    "Temperature:",
    round(store_factor_summary["Temperature_Correlation"].median(), 3)
)
print(
    "CPI:",
    round(store_factor_summary["CPI_Correlation"].median(), 3)
)
```

### Interpretation of Temperature and CPI Relationships

The median store-wise correlation between temperature and weekly sales is **−0.129**, indicating a weak negative linear relationship. For a typical store, higher temperatures were associated with slightly lower weekly sales. However, the relationship is weak, so temperature alone is unlikely to provide reliable sales predictions.

The median store-wise correlation between CPI and weekly sales is **+0.084**, indicating a very weak positive linear relationship. This suggests that CPI has little linear association with weekly sales for a typical store.

The relationships vary considerably across stores, with some showing stronger positive or negative correlations. These findings indicate association, not causation. Walmart should consider historical sales, holiday periods, seasonality, store-specific patterns, and economic and weather indicators together when planning inventory and forecasting demand.

```python

# ============================================================
# STORE-WISE UNEMPLOYMENT AND WEEKLY SALES ANALYSIS
# ============================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import pearsonr

unemployment_results = []

for store_id, group in walmart.groupby("Store"):

    store_data = group[["Weekly_Sales", "Unemployment"]].dropna()

    sales = store_data["Weekly_Sales"]
    unemployment = store_data["Unemployment"]

    # Calculate correlation and statistical significance
    if len(store_data) >= 3 and unemployment.nunique() > 1 and sales.nunique() > 1:
        correlation, p_value = pearsonr(sales, unemployment)
    else:
        correlation, p_value = np.nan, np.nan

    unemployment_results.append({
        "Store": store_id,
        "Observations": len(store_data),
        "Average_Unemployment": unemployment.mean(),
        "Average_Weekly_Sales": sales.mean(),
        "Unemployment_Sales_Correlation": correlation,
        "P_Value": p_value
    })

unemployment_summary = pd.DataFrame(unemployment_results)

# Rank stores by the strength of their negative correlation
unemployment_ranked = unemployment_summary.sort_values(
    "Unemployment_Sales_Correlation",
    ascending=True
).reset_index(drop=True)

unemployment_ranked["Negative_Correlation_Rank"] = (
    unemployment_ranked.index + 1
)

print("TOP 10 STORES WITH THE MOST NEGATIVE UNEMPLOYMENT-SALES CORRELATION")
display(unemployment_ranked.head(10).round(4))

print("\nTOP 10 STORES WITH THE MOST POSITIVE CORRELATION")
display(
    unemployment_summary.sort_values(
        "Unemployment_Sales_Correlation",
        ascending=False
    ).head(10).round(4)
)

# Visualize the ten most negative correlations
plot_data = unemployment_ranked.head(10).sort_values(
    "Unemployment_Sales_Correlation"
)

plt.figure(figsize=(11, 6))
plt.barh(
    plot_data["Store"].astype(str),
    plot_data["Unemployment_Sales_Correlation"]
)
plt.axvline(0, linestyle="--")
plt.title("Stores with the Most Negative Unemployment-Sales Correlations")
plt.xlabel("Pearson Correlation")
plt.ylabel("Store")
plt.tight_layout()
plt.show()

# Save the complete results for all stores
unemployment_summary.to_csv(
    "walmart_store_unemployment_analysis.csv",
    index=False
)

print("\nOverall median store-wise correlation:",
      round(unemployment_summary[
          "Unemployment_Sales_Correlation"
      ].median(), 4))

print("Results saved to walmart_store_unemployment_analysis.csv")
```

```python
from statsmodels.tsa.holtwinters import ExponentialSmoothing
from sklearn.metrics import mean_absolute_error, mean_squared_error
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# Select Store 20 and prepare weekly sales series
store_id = 20

store_data = (
    walmart[walmart["Store"] == store_id]
    .sort_values("Date")
    .set_index("Date")["Weekly_Sales"]
    .asfreq("7D")
)

print("Store:", store_id)
print("Number of observations:", len(store_data))
print("Date range:", store_data.index.min(), "to", store_data.index.max())
print("Missing weekly values:", store_data.isna().sum())

# Chronological train/test split
train = store_data.iloc[:-12]
test = store_data.iloc[-12:]

print("\nTraining observations:", len(train))
print("Testing observations:", len(test))

# --------------------------------------------
# Model 1: Seasonal-naive baseline
# Use sales from 52 weeks earlier when available
# --------------------------------------------

seasonal_naive_forecast = store_data.shift(52).iloc[-12:]

# --------------------------------------------
# Model 2: Holt-Winters Exponential Smoothing
# Weekly data with annual seasonality
# --------------------------------------------

hw_model = ExponentialSmoothing(
    train,
    trend="add",
    seasonal="add",
    seasonal_periods=52,
    initialization_method="estimated"
)

hw_fit = hw_model.fit(optimized=True)
hw_forecast = hw_fit.forecast(12)

# --------------------------------------------
# Evaluate both models
# --------------------------------------------

results = []

for model_name, predictions in [
    ("Seasonal Naive", seasonal_naive_forecast),
    ("Holt-Winters", hw_forecast)
]:
    predictions = pd.Series(
        np.asarray(predictions),
        index=test.index
    )

    results.append({
        "Model": model_name,
        "MAE": mean_absolute_error(test, predictions),
        "RMSE": np.sqrt(mean_squared_error(test, predictions))
    })

model_results = pd.DataFrame(results)

print("\nMODEL COMPARISON - STORE 20")
display(model_results.round(2))

# --------------------------------------------
# Plot actual vs predicted sales
# --------------------------------------------

plt.figure(figsize=(13, 5))

plt.plot(
    train.index[-30:],
    train.iloc[-30:],
    label="Training Sales"
)

plt.plot(
    test.index,
    test,
    label="Actual Test Sales",
    marker="o"
)

plt.plot(
    test.index,
    hw_forecast,
    label="Holt-Winters Forecast",
    marker="o"
)

plt.title("Store 20: Actual vs Forecast Weekly Sales")
plt.xlabel("Date")
plt.ylabel("Weekly Sales")
plt.legend()
plt.tight_layout()
plt.show()
```

```python
from statsmodels.tsa.holtwinters import ExponentialSmoothing
from sklearn.metrics import mean_absolute_error, mean_squared_error
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# Use the cleaned dataset and sort chronologically
forecast_data = walmart.sort_values(["Store", "Date"]).copy()

HORIZON = 12
SEASON_LENGTH = 52

evaluation_results = []
future_forecasts = []

for store_id, group in forecast_data.groupby("Store"):

    # Build a regular weekly series
    series = (
        group.sort_values("Date")
        .set_index("Date")["Weekly_Sales"]
        .asfreq("7D")
    )

    if series.isna().any():
        print(f"Skipping Store {store_id}: missing weekly values")
        continue

    # Chronological train/test split
    train = series.iloc[:-HORIZON]
    test = series.iloc[-HORIZON:]

    # ----------------------------------------
    # Model 1: Seasonal-naive baseline
    # ----------------------------------------
    naive_predictions = series.shift(SEASON_LENGTH).loc[test.index]

    naive_mae = mean_absolute_error(test, naive_predictions)
    naive_rmse = np.sqrt(
        mean_squared_error(test, naive_predictions)
    )

    # ----------------------------------------
    # Model 2: Holt-Winters
    # ----------------------------------------
    try:
        hw_model = ExponentialSmoothing(
            train,
            trend="add",
            seasonal="add",
            seasonal_periods=SEASON_LENGTH,
            initialization_method="estimated"
        )

        hw_fit = hw_model.fit(optimized=True)
        hw_predictions = pd.Series(
            hw_fit.forecast(HORIZON).to_numpy(),
            index=test.index
        )

        hw_mae = mean_absolute_error(test, hw_predictions)
        hw_rmse = np.sqrt(
            mean_squared_error(test, hw_predictions)
        )

        best_model = (
            "Holt-Winters"
            if hw_rmse < naive_rmse
            else "Seasonal Naive"
        )

    except Exception as error:
        print(f"Store {store_id}: Holt-Winters failed - {error}")
        hw_mae = np.nan
        hw_rmse = np.nan
        best_model = "Seasonal Naive"

    evaluation_results.append({
        "Store": store_id,
        "Naive_MAE": naive_mae,
        "Naive_RMSE": naive_rmse,
        "HoltWinters_MAE": hw_mae,
        "HoltWinters_RMSE": hw_rmse,
        "Selected_Model": best_model
    })

    # ----------------------------------------
    # Refit selected model on all history
    # and forecast the next 12 weeks
    # ----------------------------------------
    future_dates = pd.date_range(
        start=series.index[-1] + pd.Timedelta(days=7),
        periods=HORIZON,
        freq="7D"
    )

    if best_model == "Holt-Winters":
        final_model = ExponentialSmoothing(
            series,
            trend="add",
            seasonal="add",
            seasonal_periods=SEASON_LENGTH,
            initialization_method="estimated"
        ).fit(optimized=True)

        predictions = final_model.forecast(HORIZON).to_numpy()

    else:
        # Future seasonal-naive values from 52 weeks earlier
        predictions = series.iloc[-SEASON_LENGTH:-SEASON_LENGTH + HORIZON].to_numpy()

    for date, prediction in zip(future_dates, predictions):
        future_forecasts.append({
            "Store": store_id,
            "Date": date,
            "Forecast_Weekly_Sales": max(0, float(prediction)),
            "Selected_Model": best_model
        })

# --------------------------------------------
# Results and model comparison
# --------------------------------------------
evaluation_df = pd.DataFrame(evaluation_results)
forecast_df = pd.DataFrame(future_forecasts)

print("MODEL SELECTION COUNTS")
display(
    evaluation_df["Selected_Model"]
    .value_counts()
    .to_frame("Number_of_Stores")
)

print("\nAVERAGE MODEL PERFORMANCE ACROSS STORES")
display(
    evaluation_df[
        ["Naive_MAE", "Naive_RMSE",
         "HoltWinters_MAE", "HoltWinters_RMSE"]
    ].mean().to_frame("Average_Error").round(2)
)

print("\nMODEL EVALUATION - FIRST 10 STORES")
display(evaluation_df.head(10).round(2))

print("\nFORECAST PREVIEW")
display(forecast_df.head(15).round(2))

print("\nTotal forecast rows:", len(forecast_df))
print("Stores forecasted:", forecast_df["Store"].nunique())
print("Forecast date range:",
      forecast_df["Date"].min().date(), "to",
      forecast_df["Date"].max().date())

# Save results
evaluation_df.to_csv("walmart_model_evaluation.csv", index=False)
forecast_df.to_csv("walmart_12_week_forecasts.csv", index=False)

print("\nSaved:")
print("- walmart_model_evaluation.csv")
print("- walmart_12_week_forecasts.csv")
```

```python

# ============================================================
# 12. ARIMA FORECASTING FOR FIVE SELECTED WALMART STORES
# ============================================================

import warnings
warnings.filterwarnings("ignore")

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from statsmodels.tsa.arima.model import ARIMA
from sklearn.metrics import mean_absolute_error, mean_squared_error

# Select five individual stores for detailed forecasting
selected_stores = [1, 5, 10, 20, 33]

HORIZON = 12
arima_evaluation_results = []
arima_future_forecasts = []

for store_id in selected_stores:

    print("\n" + "=" * 60)
    print(f"ARIMA ANALYSIS — STORE {store_id}")
    print("=" * 60)

    # Prepare the weekly sales time series for this store
    store_series = (
        walmart.loc[walmart["Store"] == store_id]
        .sort_values("Date")
        .set_index("Date")["Weekly_Sales"]
        .asfreq("7D")
    )

    # Validate the time series
    if store_series.isna().any():
        raise ValueError(
            f"Missing weekly observations found for Store {store_id}."
        )

    if len(store_series) <= HORIZON + 20:
        raise ValueError(
            f"Insufficient observations for Store {store_id}."
        )

    # Chronological train/test split: last 12 weeks for testing
    train = store_series.iloc[:-HORIZON]
    test = store_series.iloc[-HORIZON:]

    print("Total observations:", len(store_series))
    print("Training observations:", len(train))
    print("Testing observations:", len(test))
    print("Training period:", train.index.min().date(),
          "to", train.index.max().date())
    print("Testing period:", test.index.min().date(),
          "to", test.index.max().date())

    # Fit ARIMA model using training data only
    model = ARIMA(
        train,
        order=(1, 1, 1),
        enforce_stationarity=False,
        enforce_invertibility=False
    )

    fitted_model = model.fit()

    # Predict the 12-week holdout period
    test_predictions = fitted_model.forecast(steps=HORIZON)
    test_predictions.index = test.index

    # Evaluate holdout predictions
    mae = mean_absolute_error(test, test_predictions)
    rmse = np.sqrt(mean_squared_error(test, test_predictions))

    arima_evaluation_results.append({
        "Store": store_id,
        "Model": "ARIMA(1,1,1)",
        "Training_Weeks": len(train),
        "Testing_Weeks": len(test),
        "MAE": mae,
        "RMSE": rmse
    })

    print(f"Holdout MAE:  {mae:,.2f}")
    print(f"Holdout RMSE: {rmse:,.2f}")

    # Refit ARIMA on all available observations before forecasting
    final_model = ARIMA(
        store_series,
        order=(1, 1, 1),
        enforce_stationarity=False,
        enforce_invertibility=False
    ).fit()

    future_prediction = final_model.forecast(steps=HORIZON)

    # Store future forecast values
    for forecast_date, forecast_value in future_prediction.items():
        arima_future_forecasts.append({
            "Store": store_id,
            "Date": forecast_date,
            "ARIMA_Forecast": max(0, float(forecast_value)),
            "Model": "ARIMA(1,1,1)"
        })

    # Plot historical sales, test predictions and future forecast
    plt.figure(figsize=(13, 5))

    plt.plot(
        train.index, train.values,
        label="Training Sales"
    )
    plt.plot(
        test.index, test.values,
        label="Actual Test Sales",
        marker="o"
    )
    plt.plot(
        test_predictions.index, test_predictions.values,
        label="ARIMA Test Predictions",
        marker="x"
    )
    plt.plot(
        future_prediction.index, future_prediction.values,
        label="12-Week Future Forecast",
        marker="s",
        linestyle="--"
    )

    plt.axvline(
        test.index[0],
        linestyle=":",
        label="Test Period Start"
    )

    plt.title(f"Store {store_id}: ARIMA(1,1,1) Sales Forecast")
    plt.xlabel("Date")
    plt.ylabel("Weekly Sales")
    plt.legend()
    plt.xticks(rotation=45)
    plt.tight_layout()
    plt.show()


# Create evaluation and forecast tables
arima_evaluation_df = pd.DataFrame(arima_evaluation_results)
arima_forecast_df = pd.DataFrame(arima_future_forecasts)

print("\nARIMA HOLDOUT MODEL EVALUATION")
display(arima_evaluation_df.round(2))

print("\n12-WEEK ARIMA FORECASTS FOR FIVE STORES")
display(arima_forecast_df.round(2))

print("\nAVERAGE ARIMA PERFORMANCE ACROSS THE FIVE STORES")
print("Mean MAE:", round(arima_evaluation_df["MAE"].mean(), 2))
print("Mean RMSE:", round(arima_evaluation_df["RMSE"].mean(), 2))

# Export results for the final submission
arima_evaluation_df.to_csv(
    "walmart_arima_model_evaluation.csv", index=False
)
arima_forecast_df.to_csv(
    "walmart_arima_12_week_forecasts.csv", index=False
)

print("\nSaved:")
print("- walmart_arima_model_evaluation.csv")
print("- walmart_arima_12_week_forecasts.csv")
```

## ARIMA-Based Predictive Analysis for Five Walmart Stores

### Objective
To demonstrate store-level sales forecasting, five individual stores (Stores 1, 5, 10, 20, and 33) were selected for detailed time-series analysis using the ARIMA(1,1,1) model.

### Methodology
1. Each store's weekly sales observations were separated and sorted chronologically.
2. The first 131 observations were used for training, and the final 12 historical observations were reserved for testing.
3. The ARIMA(1,1,1) model was fitted separately to each store's training data.
4. Model performance was evaluated using Mean Absolute Error (MAE) and Root Mean Squared Error (RMSE).
5. After evaluation, the model was refitted using all available historical observations to generate a 12-week forecast for each selected store.

### Results
The five-store evaluation produced an average MAE of 54,509.38 and an average RMSE of 63,601.01. Store 5 had the lowest RMSE among the selected stores, at 16,344.30, while Store 10 had the highest RMSE, at 137,183.73.

The model generated 12-week forecasts for all five stores, resulting in 60 forecast records. The visualizations compare training sales, actual test-period sales, ARIMA predictions, and future forecasts.

### Interpretation and Limitations
Forecast errors varied across stores, indicating that a single model specification may not perform equally well for every store. The forecast curves also become relatively flat over the forecast horizon, suggesting that ARIMA(1,1,1) may not fully capture recurring annual seasonality or large holiday-related sales fluctuations.

The forecasts are based on historical data ending October 26, 2012, and are illustrative historical forecasts rather than current Walmart sales predictions. Further evaluation using seasonal models such as SARIMA and comparison with alternative forecasting methods could improve model selection.

### Conclusion
This analysis demonstrates separate time-series modeling, chronological holdout evaluation, quantitative error measurement, and multi-week forecasting for five Walmart stores. The results provide a foundation for comparing forecasting methods and identifying store-specific modeling limitations.
